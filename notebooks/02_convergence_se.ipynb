# S2 — Ébauche convergence de l'erreur standard SE(N)
Objectif: vérifier que $SE \propto 1/\sqrt{N}$ (pente ~ -1/2 en log-log).

```python
import numpy as np
import matplotlib.pyplot as plt

def mc_price_se(N, seed=42, S0=100, K=100, T=1.0, r=0.05, sigma=0.2, n_steps=252):
    rng = np.random.default_rng(seed)
    dt = T/n_steps
    Z = rng.standard_normal((N, n_steps))
    S = np.exp(np.log(S0)+np.cumsum((r-0.5*sigma**2)*dt + sigma*np.sqrt(dt)*Z, axis=1))
    X = np.exp(-r*T)*np.maximum(S.mean(axis=1)-K, 0.0)
    return X.mean(), X.std(ddof=1)/np.sqrt(N)

Ns = [1000, 5000, 10000, 50000, 100000]
ses = [mc_price_se(N)[1] for N in Ns]
plt.loglog(Ns, ses, 'o-', label='SE(N)')
plt.xlabel('N'); plt.ylabel('SE'); plt.legend(); plt.grid(True, which='both'); plt.show()
```
