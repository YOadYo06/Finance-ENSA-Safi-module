Objectifs: simulation GBM sous Q, pricing asiatique call, IC 95%, reproductibilité (seed=42).

```python
import numpy as np
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
```

## Paramètres

```python
S0, K, T, r, sigma = 100.0, 100.0, 1.0, 0.05, 0.2
n_steps, N, seed = 252, 100_000, 42
dt = T / n_steps
```

## Simulation GBM et pricing asiatique

```python
rng = np.random.default_rng(seed)
Z = rng.standard_normal((N, n_steps))
increments = (r - 0.5*sigma**2)*dt + sigma*np.sqrt(dt)*Z
logS = np.log(S0) + np.cumsum(increments, axis=1)
S = np.exp(logS)
A = S.mean(axis=1)
payoffs = np.maximum(A - K, 0.0)
X = np.exp(-r*T)*payoffs
price = X.mean()
se = X.std(ddof=1)/np.sqrt(N)
ci = (price - 1.96*se, price + 1.96*se)
print({'price': price, 'se': se, 'ci95': ci})
```

## Visualisation (exemple de trajectoires)

```python
for i in range(10):
    plt.plot(S[i], alpha=0.7)
plt.title('Exemples de trajectoires GBM')
plt.xlabel('Pas de temps')
plt.ylabel('S_t')
plt.show()
```
